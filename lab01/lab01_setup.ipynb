## In-Lab Exercise 2 - columns with missing values (from info())

```python
missing = df.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

Three columns contain missing values: **deck** (688), **age** (177) and **embarked / embark_town** (2 each; embark_town is the text version of embarked).

## In-Lab Exercise 3 - largest std/mean ratio (from describe())

```python
d = df.describe().T
d["std_over_mean"] = d["std"] / d["mean"]
d[["mean","std","std_over_mean"]].sort_values("std_over_mean", ascending=False)
```

**Answer:** `sibsp` and `parch` have the largest ratio (about 2.1), then `fare` (about 1.5). A large ratio means the values are very spread out relative to their average, and columns such as `fare` (0 to 512) are on a much bigger scale than `age` or `pclass`. A model that is not given scaled inputs (e.g. gradient descent, KNN, SVM) will be dominated by the large-scale columns, so training is slow or biased.
